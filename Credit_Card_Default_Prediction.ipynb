# Day 6 - Feature Engineering and Loan-to-Income Analysis

## Objective

The objective of Day 6 is to verify the existing Loan to Income feature and create useful derived features for further analysis.

Feature engineering helps transform existing data into meaningful variables that can improve data analysis and may help machine learning models.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Credit%20Default.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Columns in dataset:")
print(df.columns.tolist())
```

```python
df['Calculated Loan to Income'] = df['Loan'] / df['Income']

df[['Income', 'Loan', 'Loan to Income', 'Calculated Loan to Income']].head(10)
```

```python
difference = abs(
    df['Loan to Income'] - df['Calculated Loan to Income']
)

print("Maximum difference:",
      difference.max())

print("Average difference:",
      difference.mean())
```

```python
if difference.max() < 0.0001:
    print("Loan to Income feature is correctly calculated.")
else:
    print("There are differences in the Loan to Income calculation.")
```

```python
df.drop('Calculated Loan to Income', axis=1, inplace=True)

print(df.columns)
```

```python
def loan_to_income_category(ratio):
    if ratio < 0.05:
        return 'Low'
    elif ratio < 0.10:
        return 'Medium'
    else:
        return 'High'

df['Loan to Income Category'] = df['Loan to Income'].apply(
    loan_to_income_category
)

df[['Loan to Income', 'Loan to Income Category']].head(10)
```

```python
df['Loan to Income Category'].value_counts()
```

```python
print("Average Loan-to-Income Ratio:", df['Loan to Income'].mean())
print("Median Loan-to-Income Ratio:", df['Loan to Income'].median())
print("Minimum Ratio:", df['Loan to Income'].min())
print("Maximum Ratio:", df['Loan to Income'].max())
print("Standard Deviation:", df['Loan to Income'].std())
```

```python
df['Loan to Income'].describe()
```

```python
Q1 = df['Loan to Income'].quantile(0.25)
Q2 = df['Loan to Income'].quantile(0.50)
Q3 = df['Loan to Income'].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Median (Q2):", Q2)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x='Loan to Income', kde=True)

plt.title('Distribution of Loan-to-Income Ratio')
plt.xlabel('Loan-to-Income Ratio')
plt.ylabel('Number of Customers')

plt.show()
```

```python
plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x='Loan to Income')

plt.title('Boxplot of Loan-to-Income Ratio')
plt.xlabel('Loan-to-Income Ratio')

plt.show()
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

ratio_outliers = df[
    (df['Loan to Income'] < lower_bound) |
    (df['Loan to Income'] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Potential Outliers:", len(ratio_outliers))
```

```python
ratio_outliers[
    ['Income', 'Loan', 'Loan to Income', 'Default']
]
```

```python
def loan_to_income_category(ratio):
    if ratio < 0.05:
        return 'Low'
    elif ratio < 0.10:
        return 'Medium'
    else:
        return 'High'

df['Loan to Income Category'] = (
    df['Loan to Income'].apply(loan_to_income_category)
)

df[['Loan to Income', 'Loan to Income Category']].head(10)
```

```python
print("Customer counts:")
print(df['Loan to Income Category'].value_counts())

print("\nCustomer percentages:")
print(
    (df['Loan to Income Category'].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Loan to Income Category')

plt.title('Customer Distribution by Loan-to-Income Category')
plt.xlabel('Loan-to-Income Category')
plt.ylabel('Number of Customers')

plt.show()
```

```python
ratio_by_default = df.groupby('Default')['Loan to Income'].agg(
    ['mean', 'median', 'std', 'min', 'max']
)

ratio_by_default
```

# Day 13 - Default Analysis by Customer Features

## Objective
To compare Income, Age, Loan Amount, and Loan-to-Income Ratio between customers who default and customers who do not default.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Credit%20Default.csv"

df = pd.read_csv(url)

print("Dataset shape:", df.shape)
print("\nFirst five rows:")
display(df.head())

print("\nDefault counts:")
print(df['Default'].value_counts())
```

```python
features = ['Income', 'Age', 'Loan', 'Loan to Income']

comparison = df.groupby('Default')[features].mean().round(2)

comparison.index = comparison.index.map({
    0: 'No Default',
    1: 'Default'
})

print("Average customer features by default status:")
display(comparison)
```

```python
features = ['Income', 'Age', 'Loan', 'Loan to Income']

for feature in features:
    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=df,
        x='Default',
        y=feature
    )

    plt.title(f'{feature} vs Default Status')
    plt.xlabel('Default (0 = No Default, 1 = Default)')
    plt.ylabel(feature)
    plt.tight_layout()
    plt.show()
```

```python
df['Income Group'] = pd.cut(
    df['Income'],
    bins=[-float('inf'), 30000, 60000, float('inf')],
    labels=['Low Income', 'Medium Income', 'High Income']
)

df['Age Group'] = pd.cut(
    df['Age'],
    bins=[-float('inf'), 30, 50, float('inf')],
    labels=['Young', 'Middle Age', 'Older']
)

df['Loan Group'] = pd.cut(
    df['Loan'],
    bins=[-float('inf'), 3000, 6000, float('inf')],
    labels=['Low Loan', 'Medium Loan', 'High Loan']
)

df['Loan to Income Group'] = pd.cut(
    df['Loan to Income'],
    bins=[-float('inf'), 0.05, 0.10, float('inf')],
    labels=['Low Ratio', 'Medium Ratio', 'High Ratio']
)

group_columns = [
    'Income Group',
    'Age Group',
    'Loan Group',
    'Loan to Income Group'
]

for group in group_columns:
    print(f"\nDefault rate by {group}:")
    rates = df.groupby(group, observed=False)['Default'].mean() * 100
    display(rates.round(2).rename('Default Rate (%)').to_frame())
```

# Day 16 - Train-Test Split

## Objective
To divide the features and target into training and testing datasets for machine learning.

## Concepts
- Training data
- Testing data
- train_test_split()
- Random state
- Stratified sampling

```python
import pandas as pd
from sklearn.model_selection import train_test_split

url = "https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Credit%20Default.csv"

df = pd.read_csv(url)

# Input features
X = df[['Income', 'Age', 'Loan', 'Loan to Income']]

# Target variable
y = df['Default']

print("Dataset loaded successfully!")
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training and testing split completed!")
print("------------------------------------")
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
print("Default distribution in original dataset:")
print(y.value_counts(normalize=True).mul(100).round(2))

print("\nDefault distribution in training data:")
print(y_train.value_counts(normalize=True).mul(100).round(2))

print("\nDefault distribution in testing data:")
print(y_test.value_counts(normalize=True).mul(100).round(2))
```

```python
print("Total records:", len(df))
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

print("\nTraining percentage:", round(len(X_train) / len(df) * 100, 2), "%")
print("Testing percentage:", round(len(X_test) / len(df) * 100, 2), "%")

print("\nMissing values in training data:")
print(X_train.isnull().sum().sum())

print("\nMissing values in testing data:")
print(X_test.isnull().sum().sum())
```

3. The training dataset contains 1,600 records.

6. The training data will be used to fit the machine learning model, while the testing data will be used to evaluate its performance.

# Day 17 - Feature Scaling using StandardScaler

## Objective
To standardize the numerical input features using StandardScaler before training a machine learning model.

## Concepts
- Feature scaling
- Standardization
- Mean and standard deviation
- StandardScaler
- Data leakage prevention

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

url = "https://raw.githubusercontent.com/YBIFoundation/Dataset/main/Credit%20Default.csv"

df = pd.read_csv(url)

X = df[['Income', 'Age', 'Loan', 'Loan to Income']]
y = df['Default']

print("Dataset loaded successfully!")
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)
```

```python
scaler = StandardScaler()

# Learn scaling parameters from training data only
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same parameters to testing data
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")
print("Scaled training shape:", X_train_scaled.shape)
print("Scaled testing shape:", X_test_scaled.shape)
```

```python
X_train_scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X.columns,
    index=X_train.index
)

print("First five rows of scaled training data:")
display(X_train_scaled_df.head())
```

```python
print("Original training data statistics:")
display(X_train.describe().round(2))

print("\nScaled training data statistics:")
display(X_train_scaled_df.describe().round(3))
```

```python
comparison = pd.DataFrame({
    'Original Mean': X_train.mean(),
    'Scaled Mean': X_train_scaled_df.mean(),
    'Original Std': X_train.std(),
    'Scaled Std': X_train_scaled_df.std()
})

display(comparison.round(3))

comparison[['Original Mean', 'Scaled Mean']].plot(
    kind='bar',
    figsize=(10, 5),
    title='Feature Means Before and After Scaling'
)

plt = __import__('matplotlib.pyplot', fromlist=['pyplot'])
plt.axhline(0, linewidth=1)
plt.ylabel('Mean')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Day 17 - Observations and Conclusion

### Observations

1. The original features have different numerical scales.
2. StandardScaler was used to standardize the input features.
3. The scaler was fitted only on the training data.
4. The same scaling parameters were applied to the testing data.
5. The scaled training features have means approximately equal to zero and standard deviations approximately equal to one.

### Conclusion

Feature scaling prepares the numerical data for machine learning algorithms that are sensitive to feature scales. The scaled training and testing datasets are now ready for the next stage of the Credit Card Default Prediction project.

# Day 18 - Logistic Regression Model

## Objective
To train a Logistic Regression model to predict credit card default using customer financial and demographic features.

## Concepts
- Logistic Regression
- Binary classification
- Model training
- Feature scaling
- Prediction

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

url = "https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Credit%20Default.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
X = df[['Income', 'Age', 'Loan', 'Loan to Income']]
y = df['Default']

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
```

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature scaling completed!")
```

```python
model = LogisticRegression(max_iter=1000)

model.fit(X_train_scaled, y_train)

print("Logistic Regression model trained successfully!")
```

```python
y_pred = model.predict(X_test_scaled)

results = pd.DataFrame({
    'Actual Default': y_test.to_numpy(),
    'Predicted Default': y_pred
})

print("First 15 actual and predicted results:")
display(results.head(15))
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Logistic Regression Accuracy:", round(accuracy * 100, 2), "%")
```

### Observations

1. Logistic Regression was selected as the first classification model.
2. The dataset was divided into training and testing sets.
3. StandardScaler was used to scale the input features.
4. The model was trained using the training dataset.
5. Predictions were generated for the testing dataset.

# Day 22: Model Comparison and Selection

## Objective
To compare Logistic Regression, Decision Tree, and Random Forest models using Accuracy, Precision, Recall, and F1-score, and identify the most suitable model for credit card default prediction.

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Load dataset
url = "https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Credit%20Default.csv"
df = pd.read_csv(url)

# Select features and target
X = df[['Income', 'Age', 'Loan', 'Loan to Income']]
y = df['Default']

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Dataset shape:", df.shape)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
# 1. Logistic Regression
# Scale the features for Logistic Regression
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train_scaled, y_train)

lr_predictions = lr_model.predict(X_test_scaled)


# 2. Decision Tree
dt_model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    random_state=42
)

dt_model.fit(X_train, y_train)
dt_predictions = dt_model.predict(X_test)


# 3. Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=42,
    class_weight='balanced'
)

rf_model.fit(X_train, y_train)
rf_predictions = rf_model.predict(X_test)

print("All three models have been trained successfully.")
```

```python
# Function to calculate model evaluation metrics
def evaluate_model(model_name, actual, predicted):
    return {
        'Model': model_name,
        'Accuracy': accuracy_score(actual, predicted),
        'Precision': precision_score(
            actual, predicted, zero_division=0
        ),
        'Recall': recall_score(
            actual, predicted, zero_division=0
        ),
        'F1-Score': f1_score(
            actual, predicted, zero_division=0
        )
    }


# Create comparison table
results = [
    evaluate_model(
        'Logistic Regression', y_test, lr_predictions
    ),
    evaluate_model(
        'Decision Tree', y_test, dt_predictions
    ),
    evaluate_model(
        'Random Forest', y_test, rf_predictions
    )
]

comparison_df = pd.DataFrame(results)

print("MODEL PERFORMANCE COMPARISON")
display(comparison_df.round(4))
```

# Day 23: ROC Curve and ROC-AUC

## Objective
To evaluate the performance of Logistic Regression, Decision Tree, and Random Forest using ROC curves and ROC-AUC scores.

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_curve, roc_auc_score

# Load dataset
url = "https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Credit%20Default.csv"
df = pd.read_csv(url)

# Select features and target
X = df[['Income', 'Age', 'Loan', 'Loan to Income']]
y = df['Default']

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Dataset shape:", df.shape)
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
# Logistic Regression: scale the features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

lr_model = LogisticRegression(max_iter=1000)
lr_model.fit(X_train_scaled, y_train)


# Decision Tree
dt_model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    random_state=42
)

dt_model.fit(X_train, y_train)


# Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=42,
    class_weight='balanced'
)

rf_model.fit(X_train, y_train)

print("All three models trained successfully.")
```

```python
# Get probabilities of class 1 (default)
lr_probabilities = lr_model.predict_proba(X_test_scaled)[:, 1]
dt_probabilities = dt_model.predict_proba(X_test)[:, 1]
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]

# Calculate ROC-AUC
roc_results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Decision Tree',
        'Random Forest'
    ],
    'ROC-AUC': [
        roc_auc_score(y_test, lr_probabilities),
        roc_auc_score(y_test, dt_probabilities),
        roc_auc_score(y_test, rf_probabilities)
    ]
})

roc_results = roc_results.sort_values(
    by='ROC-AUC',
    ascending=False
).reset_index(drop=True)

display(roc_results.round(4))

print("Highest ROC-AUC model:", roc_results.loc[0, 'Model'])
print("ROC-AUC score:", round(roc_results.loc[0, 'ROC-AUC'], 4))
```

```python
# Calculate ROC curve points
lr_fpr, lr_tpr, _ = roc_curve(y_test, lr_probabilities)
dt_fpr, dt_tpr, _ = roc_curve(y_test, dt_probabilities)
rf_fpr, rf_tpr, _ = roc_curve(y_test, rf_probabilities)

# Plot all three ROC curves
plt.figure(figsize=(9, 6))

plt.plot(
    lr_fpr, lr_tpr,
    label=f"Logistic Regression (AUC = {roc_auc_score(y_test, lr_probabilities):.3f})"
)

plt.plot(
    dt_fpr, dt_tpr,
    label=f"Decision Tree (AUC = {roc_auc_score(y_test, dt_probabilities):.3f})"
)

plt.plot(
    rf_fpr, rf_tpr,
    label=f"Random Forest (AUC = {roc_auc_score(y_test, rf_probabilities):.3f})"
)

# Random-classifier reference line
plt.plot([0, 1], [0, 1], linestyle='--', label='Random Classifier')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison')
plt.legend(loc='lower right')
plt.grid(True)
plt.tight_layout()
plt.show()
```
