# Day 3 - Data Cleaning and Data Quality

## Objective

The objective of Day 3 is to identify and handle data-quality issues
in the Credit Card Default dataset.

The dataset will be checked for missing values, duplicate records,
invalid values, incorrect data types, and unrealistic numerical values.

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Credit%20Default.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
```

```python
df.head()
```

```python
df_original = df.copy()

print("Original dataset backup created.")
```

```python
print("Rows before cleaning:", df.shape[0])
print("Columns before cleaning:", df.shape[1])
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print(missing_percentage)
```

```python
missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Percentage': (df.isnull().sum() / len(df)) * 100
})

missing_summary
```

```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
if df.duplicated().sum() > 0:
    df = df.drop_duplicates()
    print("Duplicate rows removed.")
else:
    print("No duplicate rows found.")
```

```python
print(df.dtypes)
```

```python
for column in df.columns:
    print(column, ":", df[column].dtype)
```

```python
print(df['Income'].head())
print(df['Age'].head())
print(df['Loan'].head())
print(df['Loan to Income'].head())
print(df['Default'].head())
```

```python
print("Negative Income:", (df['Income'] < 0).sum())
print("Negative Age:", (df['Age'] < 0).sum())
print("Negative Loan:", (df['Loan'] < 0).sum())
print("Negative Loan to Income:", (df['Loan to Income'] < 0).sum())
```
