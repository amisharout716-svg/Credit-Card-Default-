### Objective
The objective of this project is to analyze customer financial and demographic
information and develop a machine learning model to predict whether a customer
is likely to default on a credit card/loan-related obligation.

### Dataset Features
- Income
- Age
- Loan
- Loan to Income
- Default

## Problem Statement

Financial institutions need to identify customers who may have a higher
probability of defaulting on their credit obligations.

In this project, customer information such as income, age, loan amount,
and loan-to-income ratio will be analyzed to develop a classification model
that predicts the default status of a customer.

## Project Objectives

1. Understand and clean the credit card default dataset.
2. Perform exploratory data analysis.
3. Analyze relationships between customer attributes and default status.
4. Perform feature engineering and data preprocessing.
5. Build machine learning classification models.
6. Compare model performance using appropriate evaluation metrics.
7. Identify important features related to default prediction.
8. Generate useful business insights from the analysis.
