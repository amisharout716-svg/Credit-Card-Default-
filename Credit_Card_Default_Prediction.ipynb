```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
url = "https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Credit%20Default.csv"

df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
numeric_columns = [
    'Income',
    'Age',
    'Loan',
    'Loan to Income'
]

print(numeric_columns)
```

```python
df[numeric_columns].describe()
```

```python
mean_values = df[numeric_columns].mean()

print("Mean Values:")
print(mean_values)
```

```python
median_values = df[numeric_columns].median()

print("Median Values:")
print(median_values)
```

```python
mode_values = df[numeric_columns].mode().iloc[0]

print("Mode Values:")
print(mode_values)
```

```python
statistics = pd.DataFrame({
    'Mean': df[numeric_columns].mean(),
    'Median': df[numeric_columns].median(),
    'Standard Deviation': df[numeric_columns].std(),
    'Minimum': df[numeric_columns].min(),
    'Maximum': df[numeric_columns].max()
})

statistics
```

```python
mean_median = pd.DataFrame({
    'Mean': df[numeric_columns].mean(),
    'Median': df[numeric_columns].median()
})

mean_median
```

```python
quartiles = df[numeric_columns].quantile([0.25, 0.50, 0.75])

quartiles
```

```python
Q1 = df[numeric_columns].quantile(0.25)
Q3 = df[numeric_columns].quantile(0.75)

IQR = Q3 - Q1

print("Q1:")
print(Q1)

print("\nQ3:")
print(Q3)

print("\nIQR:")
print(IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:")
print(lower_bound)

print("\nUpper Bound:")
print(upper_bound)
```

```python
outlier_counts = {}

for column in numeric_columns:
    outliers = df[
        (df[column] < lower_bound[column]) |
        (df[column] > upper_bound[column])
    ]

    outlier_counts[column] = len(outliers)

print("Potential Outlier Counts:")
print(outlier_counts)
```

7. No outliers were removed during Day 4 because further analysis is required before deciding on outlier treatment.

```python
print("Average Income:", df['Income'].mean())
print("Average Age:", df['Age'].mean())
print("Average Loan:", df['Loan'].mean())
print("Average Loan to Income:", df['Loan to Income'].mean())
```

```python
df[['Income', 'Age', 'Loan', 'Loan to Income']].std()
```

```python
Q1 = df[['Income', 'Age', 'Loan', 'Loan to Income']].quantile(0.25)
Q3 = df[['Income', 'Age', 'Loan', 'Loan to Income']].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

for column in ['Income', 'Age', 'Loan', 'Loan to Income']:
    count = ((df[column] < lower[column]) |
             (df[column] > upper[column])).sum()
    print(column, ":", count)
```

The statistical analysis showed that the average income is 45331.6600, the average age is 40.9271 years, and the average loan amount is 4444.3697. The average loan-to-income ratio is 0.0984.
